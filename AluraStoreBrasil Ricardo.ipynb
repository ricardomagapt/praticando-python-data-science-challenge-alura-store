```python
# Importando bibliotecas
import pandas as pd
import matplotlib.pyplot as plt

# URLs dos dados das lojas
url1 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

# Lendo os dados
loja1 = pd.read_csv(url1)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

# Calculando o faturamento total de cada loja
faturamento = {
    'Loja 1': loja1['Preço'].sum(),
    'Loja 2': loja2['Preço'].sum(),
    'Loja 3': loja3['Preço'].sum(),
    'Loja 4': loja4['Preço'].sum()
}

# Convertendo para DataFrame e ordenando
faturamento_df = pd.DataFrame(list(faturamento.items()), columns=['Loja', 'Faturamento Total (R$)'])
faturamento_df = faturamento_df.sort_values(by='Faturamento Total (R$)', ascending=False)

# Exibindo a tabela de faturamento
print(faturamento_df)

# Criando o gráfico de barras
plt.figure(figsize=(8, 5))
plt.bar(faturamento_df['Loja'], faturamento_df['Faturamento Total (R$)'])
plt.title('Faturamento Total por Loja')
plt.xlabel('Loja')
plt.ylabel('Faturamento Total (R$)')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

```

# 2. Vendas por Categoria

```python

# Agrupando os dados de todas as lojas
todas_as_lojas = pd.concat([loja1, loja2, loja3, loja4])

# Contando quantos produtos foram vendidos por categoria
vendas_por_categoria = todas_as_lojas['Categoria do Produto'].value_counts().reset_index()
vendas_por_categoria.columns = ['Categoria', 'Quantidade de Vendas']

# Exibindo a tabela
print(vendas_por_categoria)

# Criando um gráfico de barras
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.bar(vendas_por_categoria['Categoria'], vendas_por_categoria['Quantidade de Vendas'])
plt.title('Vendas por Categoria de Produto (Todas as Lojas)')
plt.xlabel('Categoria')
plt.ylabel('Quantidade de Vendas')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

# 3. Média de Avaliação das Lojas


```python
# Calculando a média de avaliação de cada loja
avaliacoes = {
    'Loja 1': loja1['Avaliação da compra'].mean(),
    'Loja 2': loja2['Avaliação da compra'].mean(),
    'Loja 3': loja3['Avaliação da compra'].mean(),
    'Loja 4': loja4['Avaliação da compra'].mean()
}

# Convertendo em DataFrame
avaliacoes_df = pd.DataFrame(list(avaliacoes.items()), columns=['Loja', 'Média de Avaliação'])
avaliacoes_df = avaliacoes_df.sort_values(by='Média de Avaliação', ascending=False)

# Exibindo a tabela
print(avaliacoes_df)

# Criando gráfico de barras
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.bar(avaliacoes_df['Loja'], avaliacoes_df['Média de Avaliação'])
plt.title('Média de Avaliação por Loja')
plt.xlabel('Loja')
plt.ylabel('Média de Avaliação (1 a 5)')
plt.ylim(0, 5)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

# 5. Frete Médio por Loja

```python
import matplotlib.pyplot as plt

# Função para gerar gráficos dos top 5 e bottom 5 produtos por loja
def graficos_top_bottom(df, nome_loja):
    contagem = df['Produto'].value_counts()

    top_5 = contagem.head(5)
    bottom_5 = contagem.tail(5)

    # Gráfico Top 5
    plt.figure(figsize=(8, 4))
    plt.bar(top_5.index, top_5.values)
    plt.title(f'{nome_loja} - Top 5 Produtos Mais Vendidos')
    plt.xlabel('Produto')
    plt.ylabel('Quantidade Vendida')
    plt.xticks(rotation=45, ha='right')
    plt.grid(axis='y', linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.show()

    # Gráfico Bottom 5
    plt.figure(figsize=(8, 4))
    plt.bar(bottom_5.index, bottom_5.values)
    plt.title(f'{nome_loja} - Top 5 Produtos Menos Vendidos')
    plt.xlabel('Produto')
    plt.ylabel('Quantidade Vendida')
    plt.xticks(rotation=45, ha='right')
    plt.grid(axis='y', linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.show()

# Gerar os gráficos para cada loja
graficos_top_bottom(loja1, 'Loja 1')
graficos_top_bottom(loja2, 'Loja 2')
graficos_top_bottom(loja3, 'Loja 3')
graficos_top_bottom(loja4, 'Loja 4')
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Calculando o frete médio por loja
fretes = {
    'Loja 1': loja1['Frete'].mean(),
    'Loja 2': loja2['Frete'].mean(),
    'Loja 3': loja3['Frete'].mean(),
    'Loja 4': loja4['Frete'].mean()
}

# Criando DataFrame para visualização
frete_df = pd.DataFrame(list(fretes.items()), columns=['Loja', 'Frete Médio (R$)'])
frete_df = frete_df.sort_values(by='Frete Médio (R$)', ascending=True)

# Exibindo a tabela
print(frete_df)

# Criando gráfico de barras
plt.figure(figsize=(8, 5))
plt.bar(frete_df['Loja'], frete_df['Frete Médio (R$)'])
plt.title('Frete Médio por Loja')
plt.xlabel('Loja')
plt.ylabel('Frete Médio (R$)')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
